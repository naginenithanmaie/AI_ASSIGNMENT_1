###**Q4: University Exam Timetable Optimization Using Local Search**

There are N courses and S slots. We need to assign every course to one slot so that the timetable has minimum possible cost.

So the problem here is, some courses have common students.

For example, if two courses C1 and C2 have common students and if this conflicting pair is assigned to same slot a penalty will be added.

The objective is to:
Minimize Total Cost
A lower cost represents a better timetable.
The ideal solution has:
Total Cost = 0

**Module 1: Input Processing**

```python
def take_input():

    N, S = map(int, input("Enter number of courses and slots: ").split())

    K = int(input("Enter number of conflicting pairs: "))

    conflicts = []

    print("Enter conflicting course pairs:")

    for i in range(K):
        a, b = map(int, input().split())
        conflicts.append((a, b))

    print("Enter initial timetable:")
    timetable = list(map(int, input().split()))

    return N, S, conflicts, timetable


N, S, conflicts, initial_timetable = take_input()

print("\nNumber of courses:", N)
print("Number of slots:", S)
print("Conflicts:", conflicts)
print("Initial timetable:", initial_timetable)
```

**Module 2: State Representation**

```python
def display_timetable(timetable):

    for i in range(N):
        print("C" + str(i + 1), "→ Slot", timetable[i])


print("Initial Timetable:")
display_timetable(initial_timetable)
```

**Module 3: Cost Function**

The cost has two parts:

Total Cost
    =
Conflict Penalty
    +
Distribution Penalty

Conflict Penalty:
Every conflicting pair placed in the same slot gives a penalty of 10.

Distribution Penalty:
For every course beyond two courses in a single slot: Penalty = 2

Total Cost = 10 × Number of Course Conflicts + 2 × Total Number of Excess Courses


```python
def calculate_cost(timetable):

    conflict_count = 0

    for a, b in conflicts:

        if timetable[a - 1] == timetable[b - 1]:
            conflict_count += 1

    excess_courses = 0

    for slot in range(1, S + 1):

        count = timetable.count(slot)

        if count > 2:
            excess_courses += count - 2

    conflict_penalty = conflict_count * 10
    distribution_penalty = excess_courses * 2

    total_cost = conflict_penalty + distribution_penalty

    return conflict_count, conflict_penalty, excess_courses, distribution_penalty, total_cost
```

**Test cost function**

```python
def display_cost(timetable):

    result = calculate_cost(timetable)

    print("Number of course conflicts:", result[0])
    print("Conflict penalty:", result[1])
    print("Excess courses:", result[2])
    print("Distribution penalty:", result[3])
    print("Total cost:", result[4])


display_cost(initial_timetable)
```

**Module 4: Neighbor Generation**

## Neighbor Generation

A neighbouring state is created by changing the slot
of exactly one course.

For every course, we try all the other examination slots.

For example:

Current state:

[1, 1, 2, 3, 4, 2]

Changing C1 from Slot 1 to Slot 2 gives:

[2, 1, 2, 3, 4, 2]

Changing C1 from Slot 1 to Slot 3 gives:

[3, 1, 2, 3, 4, 2]

Changing C1 from Slot 1 to Slot 4 gives:

[4, 1, 2, 3, 4, 2]

For 6 courses and 4 slots:

Number of neighbours = 6 × (4 - 1)
                     = 18

Generating Neighbours

```python
def get_neighbors(timetable):

    neighbors = []

    for course in range(N):

        for slot in range(1, S + 1):

            if timetable[course] != slot:

                new_timetable = timetable.copy()

                new_timetable[course] = slot

                neighbors.append(new_timetable)

    return neighbors
```

```python
neighbors = get_neighbors(initial_timetable)

print("Number of neighbours:", len(neighbors))

for neighbor in neighbors:
    print(neighbor)
```

**Module 5: Hill Climbing**

## Hill Climbing Algorithm

The Hill Climbing algorithm works as follows:

1. Start with the initial timetable.
2. Calculate its cost.
3. Generate all neighbouring timetables.
4. Calculate the cost of every neighbour.
5. Select the neighbour with the lowest cost.
6. If multiple neighbours have the same cost, select the first one.
7. Move to the neighbour only if its cost is strictly lower.
8. Repeat the process.
9. Stop when no better neighbour exists.

The stopping condition is called a Local Optimum.

```python
def hill_climbing(start_timetable):

    current = start_timetable.copy()

    iterations = 0
    neighbors_evaluated = 0

    while True:

        current_cost = calculate_cost(current)[4]

        neighbors = get_neighbors(current)

        best_neighbor = None
        best_cost = float('inf')

        for neighbor in neighbors:

            cost = calculate_cost(neighbor)[4]

            neighbors_evaluated += 1

            if cost < best_cost:

                best_cost = cost
                best_neighbor = neighbor

        print("\nIteration", iterations + 1)
        print("Current timetable:", current)
        print("Current cost:", current_cost)
        print("Best neighbour:", best_neighbor)
        print("Best neighbour cost:", best_cost)

        if best_cost >= current_cost:

            break

        current = best_neighbor

        iterations += 1

    return current, iterations, neighbors_evaluated
```

## Run Hill Climbing

The algorithm is now applied to the initial timetable.

The program records:

- Final timetable
- Final conflict penalty
- Final distribution penalty
- Final total cost
- Number of iterations
- Number of neighbouring states evaluated
- Execution time
- Termination reason

```python
import time

start_time = time.time()

final_timetable, iterations, neighbors_evaluated = hill_climbing(
    initial_timetable
)

end_time = time.time()

execution_time = end_time - start_time

print("\n" + "=" * 40)
print("FINAL RESULT")
print("=" * 40)

print("\nFinal Timetable:")
display_timetable(final_timetable)

print("\nFinal Cost:")
display_cost(final_timetable)

print("\nIterations:", iterations)
print("Neighbours evaluated:", neighbors_evaluated)
print("Execution time:", execution_time, "seconds")
print("Termination reason: Local Optimum")
```

## Multiple Experiments

Hill Climbing can produce different results depending on
the initial timetable.

Therefore, the algorithm is tested using three different
initial timetables.

Experiment 1:

[1, 1, 2, 3, 4, 2]

Experiment 2:

[2, 3, 1, 1, 4, 4]

Experiment 3:

[4, 2, 3, 1, 2, 3]

The results of all experiments are compared.

Three Experiments


```python
experiments = [
    [1, 1, 2, 3, 4, 2],
    [2, 3, 1, 1, 4, 4],
    [4, 2, 3, 1, 2, 3]
]

results = []

for i in range(3):

    initial = experiments[i]

    start_time = time.time()

    final, iterations, neighbors_evaluated = hill_climbing(initial)

    end_time = time.time()

    initial_cost = calculate_cost(initial)[4]
    final_cost = calculate_cost(final)[4]

    result = {
        "experiment": i + 1,
        "initial": initial,
        "initial_cost": initial_cost,
        "final": final,
        "final_cost": final_cost,
        "iterations": iterations,
        "neighbors": neighbors_evaluated,
        "time": end_time - start_time
    }

    results.append(result)
```

- Initial timetable
