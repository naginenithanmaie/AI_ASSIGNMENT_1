```python
import heapq
import time
```

```python
def reconstruct_path(parent, goal):
    path = []

    while goal is not None:
        path.append(goal)
        goal = parent.get(goal)

    return path[::-1]


def calculate_path_cost(path, graph):
    total_cost = 0

    for i in range(len(path) - 1):
        current = path[i]
        next_node = path[i + 1]

        for neighbor, cost in graph[current]:
            if neighbor == next_node:
                total_cost += cost
                break

    return total_cost
```

```python
def greedy_best_first_search(graph, heuristic, start, goal):
    start_time = time.perf_counter()

    priority_queue = []
    heapq.heappush(priority_queue, (heuristic[start], start))

    visited = set()
    parent = {start: None}

    nodes_expanded = 0

    while priority_queue:
        _, current = heapq.heappop(priority_queue)

        if current in visited:
            continue

        visited.add(current)
        nodes_expanded += 1

        if current == goal:
            end_time = time.perf_counter()

            path = reconstruct_path(parent, goal)
            total_cost = calculate_path_cost(path, graph)

            return {
                "found": True,
                "path": path,
                "cost": total_cost,
                "expanded": nodes_expanded,
                "time": end_time - start_time
            }

        for neighbor, cost in graph[current]:
            if neighbor not in visited:
                if neighbor not in parent:
                    parent[neighbor] = current

                heapq.heappush(
                    priority_queue,
                    (heuristic[neighbor], neighbor)
                )

    end_time = time.perf_counter()

    return {
        "found": False,
        "path": [],
        "cost": 0,
        "expanded": nodes_expanded,
        "time": end_time - start_time
    }
```

```python
def a_star_search(graph, heuristic, start, goal):
    start_time = time.perf_counter()

    priority_queue = []

    heapq.heappush(
        priority_queue,
        (heuristic[start], 0, start)
    )

    g_cost = {start: 0}
    parent = {start: None}
    closed = set()

    nodes_expanded = 0

    while priority_queue:
        f_current, current_g, current = heapq.heappop(priority_queue)

        if current in closed:
            continue

        closed.add(current)
        nodes_expanded += 1

        if current == goal:
            end_time = time.perf_counter()

            path = reconstruct_path(parent, goal)

            return {
                "found": True,
                "path": path,
                "cost": current_g,
                "expanded": nodes_expanded,
                "time": end_time - start_time
            }

        for neighbor, edge_cost in graph[current]:
            tentative_g = current_g + edge_cost

            if neighbor not in g_cost or tentative_g < g_cost[neighbor]:
                g_cost[neighbor] = tentative_g
                parent[neighbor] = current

                f_value = tentative_g + heuristic[neighbor]

                heapq.heappush(
                    priority_queue,
                    (f_value, tentative_g, neighbor)
                )

    end_time = time.perf_counter()

    return {
        "found": False,
        "path": [],
        "cost": 0,
        "expanded": nodes_expanded,
        "time": end_time - start_time
    }
```

# Comparison of Greedy Best-First Search and A* Search

| Criteria | Greedy Best-First Search | A* Search |
|---|---|---|
| **Path Cost** | 101 | 6 |
| **Path** | A → B → F | A → C → D → F |
| **States Explored** | 3 | 5 |
| **Execution Time** | Depends on execution | Depends on execution |
| **Optimality** | Not guaranteed to be optimal | Optimal when the heuristic is admissible |
| **Heuristic Used** | Uses only `h(n)` | Uses `g(n) + h(n)` |
| **Effectiveness** | Faster decision-making but can be misled by the heuristic | More reliable because it considers both actual and estimated cost |

## Conclusion

Greedy Best-First Search explored fewer states but selected a high-cost path because it relied only on the heuristic value. A* Search explored more states but found the optimal path by considering both the actual path cost `g(n)` and the heuristic `h(n)`. Therefore, A* was more effective for finding the lowest-cost route in this test case.

Execution time may vary between runs, so very small differences in execution time should not be considered significant.
