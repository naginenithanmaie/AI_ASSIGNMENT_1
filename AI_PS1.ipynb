#Problem Statement 1:

A rescue robot navigates a grid from **S** to **G**, avoiding **#**,
moving through **.** (free cells). Successors are generated like:
**UP - RIGHT - DOWN - LEFT**.

we compare **BFS** and **DFS** for this pathfinding task.

First,  we have some helper functions to read the grid and check if a move is valid.

```python
from collections import deque
import time

def parse_grid(grid_lines):
    grid = [list(row) for row in grid_lines]
    R = len(grid)
    C = len(grid[0]) if R > 0 else 0
    start = goal = None
    for i in range(R):
        for j in range(C):
            if grid[i][j] == 'S':
                start = (i, j)
            elif grid[i][j] == 'G':
                goal = (i, j)
    return grid, R, C, start, goal


def is_valid(r, c, R, C, grid):
    return 0 <= r < R and 0 <= c < C and grid[r][c] != '#'
```

Now we decide what moves are possible from a cell in the  Order: UP, RIGHT, DOWN, LEFT.

```python
DIRECTIONS = [
    ("UP", -1, 0),
    ("RIGHT", 0, 1),
    ("DOWN", 1, 0),
    ("LEFT", 0, -1)
]

def get_successors(state, grid, R, C):
    r, c = state
    successors = []
    for action, dr, dc in DIRECTIONS:
        nr, nc = r + dr, c + dc
        if is_valid(nr, nc, R, C, grid):
            successors.append((action, (nr, nc)))
    return successors
```

BFS explores level by level, so it always finds the shortest path.

```python
def bfs(grid, R, C, start, goal):
    start_time = time.time()
    frontier = deque([start])
    visited = {start}
    parent = {start: (None, None)}
    nodes_expanded = 0

    if start == goal:
        exec_time = time.time() - start_time
        return True, [], 0, nodes_expanded, exec_time

    found = False
    while frontier:
        current = frontier.popleft()
        nodes_expanded += 1

        if current == goal:
            found = True
            break

        for action, nxt in get_successors(current, grid, R, C):
            if nxt not in visited:
                visited.add(nxt)
                parent[nxt] = (current, action)
                frontier.append(nxt)

    exec_time = time.time() - start_time

    if not found:
        return False, [], 0, nodes_expanded, exec_time


    path = []
    node = goal
    while parent[node][0] is not None:
        prev, action = parent[node]
        path.append(action)
        node = prev
    path.reverse()

    return True, path, len(path), nodes_expanded, exec_time
```

DFS goes deep down one path before backtracking. Might not be the shortest path.

```python
def dfs(grid, R, C, start, goal):
    start_time = time.time()
    stack = [start]
    visited = {start}
    parent = {start: (None, None)}
    nodes_expanded = 0

    if start == goal:
        exec_time = time.time() - start_time
        return True, [], 0, nodes_expanded, exec_time

    found = False
    while stack:
        current = stack.pop()
        nodes_expanded += 1

        if current == goal:
            found = True
            break

        successors = get_successors(current, grid, R, C)
        for action, nxt in reversed(successors):
            if nxt not in visited:
                visited.add(nxt)
                parent[nxt] = (current, action)
                stack.append(nxt)

    exec_time = time.time() - start_time

    if not found:
        return False, [], 0, nodes_expanded, exec_time

    path = []
    node = goal
    while parent[node][0] is not None:
        prev, action = parent[node]
        path.append(action)
        node = prev
    path.reverse()

    return True, path, len(path), nodes_expanded, exec_time
```

We have a helper to print results in the required format.

```python
def run_algorithm(name, algo_func, grid, R, C, start, goal):
    print(f"Algorithm: {name}")
    found, path, moves, nodes_expanded, exec_time = algo_func(grid, R, C, start, goal)

    if found:
        print("Path Found: Yes")
        print(f"Path: {' '.join(path)}")
        print(f"Number of Moves = {moves}")
        print(f"Nodes Expanded = {nodes_expanded}")
        print(f"Execution Time = {exec_time:.6f} seconds")
    else:
        print("Path Found: No")
        print(f"Nodes Expanded = {nodes_expanded}")
        print(f"Execution Time = {exec_time:.6f} seconds")
    print()
    return found, path, moves, nodes_expanded, exec_time
```

Main function that ties everything together and runs both algorithms.

```python
def solve(grid_lines):
    grid, R, C, start, goal = parse_grid(grid_lines)

    if start is None or goal is None:
        print("Error: Grid must contain exactly one S and one G.")
        return

    bfs_result = run_algorithm("BFS", bfs, grid, R, C, start, goal)
    dfs_result = run_algorithm("DFS", dfs, grid, R, C, start, goal)

    return bfs_result, dfs_result


def read_input_and_solve():
    first_line = input().split()
    R, C = int(first_line[0]), int(first_line[1])
    grid_lines = [input() for _ in range(R)]
    return solve(grid_lines)
```

```python
import pandas as pd

def compare_algorithms(grid_lines, label="Test Case"):
    grid, R, C, start, goal = parse_grid(grid_lines)

    bfs_found, bfs_path, bfs_moves, bfs_nodes, bfs_time = bfs(grid, R, C, start, goal)
    dfs_found, dfs_path, dfs_moves, dfs_nodes, dfs_time = dfs(grid, R, C, start, goal)

    data = {
        "Metric": ["Path Found", "Path Length (Moves)", "Nodes Expanded",
                   "Execution Time (s)", "Optimal?"],
        "BFS": [
            "Yes" if bfs_found else "No",
            bfs_moves if bfs_found else "-",
            bfs_nodes,
            f"{bfs_time:.6f}",
            "Yes (guaranteed shortest path)" if bfs_found else "-"
        ],
        "DFS": [
            "Yes" if dfs_found else "No",
            dfs_moves if dfs_found else "-",
            dfs_nodes,
            f"{dfs_time:.6f}",
            "Not guaranteed" if dfs_found else "-"
        ]
    }

    df = pd.DataFrame(data)
    print(f"\n=== Comparison: {label} ===")
    display(df)
    return df
```

```python
compare_df = compare_algorithms(grid_lines, "User Input Grid")
```

Comparision: BFS vs DFS                                        
 Path Length: Always shortest (optimal)in BFS whereas DFS finds a longer path (non-optimal)

 States Explored: BFS explores more nodes near the start uniformly whereas DFS explores less nodes if goal is found deep in first branch.

Execution Time: BFS has Consistent, proportional to shortest path depth whereas DFS may be faster or slower depending on traversal order.

Optimality: Guaranteed optimal in BFS but not guarantyd optimal in DFS
Memory Usage:| Higher in BFS and Lower in DFS.
